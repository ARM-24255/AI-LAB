```python
import time
import tracemalloc

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H'],
    'E': ['I'],
    'F': [],
    'G': ['J']
}

visited = set()

def dfs(node, goal):
    if node in visited:
        return False

    visited.add(node)
    print(node, end=" ")

    if node == goal:
        return True

    for neighbor in graph.get(node, []):
        if dfs(neighbor, goal):
            return True

    return False

start = input("Enter start node: ").upper()
goal = input("Enter goal node: ").upper()

tracemalloc.start()
start_time = time.perf_counter()

print("DFS Traversal:")
found = dfs(start, goal)

end_time = time.perf_counter()
current, peak = tracemalloc.get_traced_memory()
tracemalloc.stop()

print("\nGoal found!" if found else "\nGoal not found!")
print("Execution time:", round(end_time - start_time, 8), "seconds")
print("Current memory:", current, "bytes")
print("Peak memory:", peak, "bytes")
print("Theoretical time complexity: O(V + E)")
print("Theoretical space complexity: O(V)")
```

```python
import time
import tracemalloc
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

def bfs(start, goal):
    visited = {start}
    queue = deque([start])

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        if node == goal:
            return True

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return False

start = input("Enter start node: ").upper()
goal = input("Enter goal node: ").upper()

tracemalloc.start()
start_time = time.perf_counter()

print("BFS Traversal:")
found = bfs(start, goal)

end_time = time.perf_counter()
current, peak = tracemalloc.get_traced_memory()
tracemalloc.stop()

print("\nGoal found!" if found else "\nGoal not found!")
print("Execution time:", round(end_time - start_time, 8), "seconds")
print("Current memory:", current, "bytes")
print("Peak memory:", peak, "bytes")
print("Time complexity: O(V + E)")
print("Space complexity: O(V)")
```

```python
import time
import tracemalloc

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

def dls(node, goal, depth):
    print(node, end=" ")

    if node == goal:
        return True

    if depth == 0:
        return False

    for child in graph.get(node, []):
        if dls(child, goal, depth - 1):
            return True

    return False

def iddfs(start, goal):
    for depth in range(len(graph) + 1):
        print("\nDepth limit:", depth)

        if dls(start, goal, depth):
            return True

    return False

start = input("Enter start node: ").upper()
goal = input("Enter goal node: ").upper()

tracemalloc.start()
start_time = time.perf_counter()

found = iddfs(start, goal)

end_time = time.perf_counter()
current, peak = tracemalloc.get_traced_memory()
tracemalloc.stop()

print("\nGoal found!" if found else "\nGoal not found!")
print("Execution time:", round(end_time - start_time, 8), "seconds")
print("Current memory:", current, "bytes")
print("Peak memory:", peak, "bytes")
print("Time complexity: O(b^d)")
print("Space complexity: O(bd)")
```
