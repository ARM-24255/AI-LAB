```python
import heapq

def manhattan(state, goal):
    distance = 0

    for i in range(9):
        if state[i] != 0:
            j = goal.index(state[i])
            distance += abs(i // 3 - j // 3)
            distance += abs(i % 3 - j % 3)

    return distance


def solve(start, goal):
    queue = []
    heapq.heappush(queue, (manhattan(start, goal), 0, start, []))
    visited = set()

    while queue:
        f, cost, state, path = heapq.heappop(queue)

        if state in visited:
            continue

        visited.add(state)
        path = path + [state]

        if state == goal:
            print("\nSolution found!")
            for step, s in enumerate(path):
                print("\nStep", step)
                for i in range(0, 9, 3):
                    print(s[i:i+3])
            return

        zero = state.index(0)
        row, col = zero // 3, zero % 3

        for move in [-3, 3, -1, 1]:
            new_pos = zero + move

            if new_pos < 0 or new_pos > 8:
                continue
            if move == -1 and col == 0:
                continue
            if move == 1 and col == 2:
                continue

            new_state = list(state)
            new_state[zero], new_state[new_pos] = (
                new_state[new_pos], new_state[zero]
            )
            new_state = tuple(new_state)

            if new_state not in visited:
                new_cost = cost + 1
                h = manhattan(new_state, goal)
                heapq.heappush(
                    queue, (new_cost + h, new_cost, new_state, path)
                )

    print("No solution found!")


print("Enter start state (use 0 for blank):")
start = tuple(map(int, input().split()))

print("Enter goal state (use 0 for blank):")
goal = tuple(map(int, input().split()))

if len(start) == 9 and len(goal) == 9:
    if sorted(start) == list(range(9)) and sorted(goal) == list(range(9)):
        solve(start, goal)
    else:
        print("Enter numbers 0 to 8 exactly once.")
else:
    print("Enter exactly 9 numbers for each state.")
```
